```python
# Celda 1: librerías
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
```

```python
# Celda 2: cargar datos (imágenes 8x8 de dígitos)
digitos = load_digits()
x = digitos.images
y = digitos.target
print(x.shape, y.shape)
```

```python
# Celda 3: ver algunas imágenes
plt.figure(figsize=(8, 2))
for i in range(8):
    plt.subplot(1, 8, i + 1)
    plt.imshow(x[i], cmap="gray_r")
    plt.title(y[i])
    plt.axis("off")
plt.show()
```

```python
# Celda 4: preprocesamiento (normalizar y aplanar)
x = x / 16.0
x = x.reshape(-1, 64)
print(x.shape)
```

```python
# Celda 5: separar entrenamiento y prueba
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Celda 6: red neuronal
model = tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(64,)),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
```

```python
# Celda 7: entrenar
model.fit(x_train, y_train, epochs=30, validation_split=0.2)
```

```python
# Celda 8: predecir y evaluar
y_pred = model.predict(x_test).argmax(axis=1)
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
# Celda 9: matriz de confusión
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred)).plot(cmap="Blues")
plt.show()
```

```python
# Celda 10: comparar con KNN y SVM
knn = KNeighborsClassifier(n_neighbors=3).fit(x_train, y_train)
svm = SVC().fit(x_train, y_train)
print("KNN:", accuracy_score(y_test, knn.predict(x_test)))
print("SVM:", accuracy_score(y_test, svm.predict(x_test)))
```
